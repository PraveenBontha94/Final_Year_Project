```python
import pandas as pd

df = pd.read_csv("dataset/Online_Courses.csv")

# Keep only Skills column
skills_series = df["Skills"].dropna()
```

```python
print(skills_series.head())
```

```python
skills_list = []

for s in skills_series:
    skills_list.extend([x.strip().lower() for x in s.split(",")])
```

```python
unique_skills = list(set(skills_list))
print(f"Total unique skills: {len(unique_skills)}")
print(unique_skills)
```
