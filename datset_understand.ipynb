```python
import pandas as pd

# Load a local CSV file
df1 = pd.read_csv('dataset/coursea_data.csv')
df2 = pd.read_csv('dataset/Online_Courses.csv')

# Quick data checks
print('df1 shape:', df1.shape)
print('df1 columns:', df1.columns.tolist())
print('df1 sample:\n', df1.head(3).to_string(index=False))
print('df1 nulls:\n', df1.isnull().sum())

print('df2 shape:', df2.shape)
print('df2 columns:', df2.columns.tolist())
print('df2 sample:\n', df2.head(3).to_string(index=False))
print('df2 nulls:\n', df2.isnull().sum())
```

```python
print(df1.columns)
print(df2.columns)
```

```python
df = df2[[
    "Title",
    "Short Intro",
    "Skills",
    "Rating",
    "Level"
]].copy()

df.columns = ["course", "description", "skills", "rating", "level"]

df.fillna("", inplace=True)
df.drop_duplicates(inplace=True)

print('df (from df2) shape:', df.shape)
print('df columns:', df.columns.tolist())
print('df sample:\n', df.head(3).to_string(index=False))
print('df nulls:\n', df.isnull().sum())
```

```python
df["skills"] = df["skills"].apply(
    lambda x: [s.strip().lower() for s in x.split(",")] if x else []
)
print('sample skills for first 5 rows:', df['skills'].head(5).tolist())
print('total number of skill entries across rows:', df['skills'].map(len).sum())
```

```python
# Flatten skills and print unique technical skills + counts
from collections import Counter

# Ensure `df` exists and has a `skills` column of lists
if 'df' not in globals():
    raise NameError('DataFrame `df` not found. Run the earlier cells that create `df`.')

all_skills = [s for sublist in df['skills'] for s in sublist]
skill_counts = Counter(all_skills)
unique_skills = sorted(skill_counts.keys())

# Define technical keyword substrings (lowercase) to detect technical skills
technical_keywords = {
    'python','java','c++','c#','r','sql','nosql','mongodb','postgres','mysql','oracle',
    'tensorflow','keras','pytorch','torch','scikit','sklearn','numpy','pandas','matplotlib',
    'seaborn','opencv','nlp','natural language','computer vision','deep learning',
    'machine learning','ml','data engineering','data engineering','hadoop','spark','scala',
    'hive','pig','aws','azure','gcp','google cloud','docker','kubernetes','kubectl','helm',
    'terraform','ansible','ci/cd','jenkins','git','github','gitlab','html','css','javascript',
    'node','react','angular','vue','flask','django','rest','api','api development','api design',
    'apache','spark','elastic','elasticsearch','kafka','redis','rabbitmq','xgboost','lightgbm',
    'tableau','power bi','powerbi','dax','sql server','mlops','computer vision','nlp','bash','linux',
    'cassandra','hbase','spark','s3','lambda','ecs','ec2','bigquery','dataproc','snowflake'
}

def is_technical(skill):
    s = skill.lower()
    # keep if any technical keyword is a substring
    for kw in technical_keywords:
        if kw in s:
            return True
    # additional heuristics: languages or tokens like 'js', 'cpp'
    if s in {'js','cpp'}:
        return True
    return False

technical_skills = [s for s in all_skills if is_technical(s)]
non_technical_skills = [s for s in all_skills if not is_technical(s)]

tech_counts = Counter(technical_skills)
unique_technical = sorted(set(technical_skills))

print('Total distinct technical skills:', len(unique_technical))
print('\nAll unique technical skills:\n')
for skill in unique_technical:
    print(skill)

print('\nTop 50 technical skills by frequency:')
for skill, cnt in tech_counts.most_common(50):
    print(f'{skill}: {cnt}')

print('\nRemoved non-technical skills sample (first 50 distinct):')
for skill in sorted(set(non_technical_skills))[:50]:
    print(skill)
```
